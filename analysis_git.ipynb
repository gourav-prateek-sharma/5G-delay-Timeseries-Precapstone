```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the JSON file
with open("packets.json", "r") as file:
    packets = json.load(file)

# Extract Sequence Numbers and RLC Delays
sn_values = []
rlc_delays = []
outliers = []

for packet in packets[:40000]:  # Process first 40,000 packets
    try:
        rlc_delay = (packet['rlc.out_t'] - packet['rlc.in_t']) * 1e3  # Convert to milliseconds
        sn_values.append(packet["sn"])
        rlc_delays.append(rlc_delay)

        # Store outliers (delay > 40ms)
        if rlc_delay > 40:
            outliers.append(packet)

    except KeyError:
        continue  # Skip packets with missing timestamps


# Convert lists to numpy arrays
sn_values = np.array(sn_values)
rlc_delays = np.array(rlc_delays)

# 1️⃣ Plot Delay vs Sequence Number
plt.figure(figsize=(12, 6))
sns.scatterplot(x=sn_values, y=rlc_delays, color="blue", alpha=0.5)
plt.xlabel("Sequence Number (SN)")
plt.ylabel("RLC Delay (ms)")
plt.title("RLC Delay vs Sequence Number")
plt.grid(True)
plt.show()

# 2️⃣ Store Outliers in JSON and Plot Outlier Delay vs SN
with open("packets.outliers.rlc.json", "w") as out_file:
    json.dump(outliers, out_file, indent=4)

# Extract Outlier SNs and Delays
outlier_sn = [pkt["sn"] for pkt in outliers]
outlier_delays = [((pkt['rlc.out_t'] - pkt['rlc.in_t']) * 1e3) for pkt in outliers]

# Scatter Plot of Outliers
plt.figure(figsize=(12, 6))
sns.scatterplot(x=outlier_sn, y=outlier_delays, color="red", alpha=0.5)
plt.xlabel("Sequence Number (SN)")
plt.ylabel("RLC Delay (ms)")
plt.title("Outlier RLC Delay (>40ms) vs Sequence Number")
plt.grid(True)
plt.show()

# 3️⃣ Plot CCDF (Complementary CDF) of Non-Outlier Delays
filtered_delays = np.array([d for d in rlc_delays if d <= 40])  # Exclude outliers
sorted_delays = np.sort(filtered_delays)
ccdf = 1.0 - np.arange(1, len(sorted_delays) + 1) / len(sorted_delays)

plt.figure(figsize=(10, 5))
plt.plot(sorted_delays, ccdf, marker="o", linestyle="-", color="purple")
plt.xlabel("RLC Delay (ms)")
plt.ylabel("CCDF (Log Scale)")
plt.yscale("log")  # Set Y-axis to logarithmic scale
plt.title("Logarithmic CCDF of RLC Delays (Excluding Outliers)")
plt.grid(True)
plt.show()

#  Plot CCDF (Complementary CDF) of Outlier Delays
filtered_delays = np.array([d for d in rlc_delays if d >= 40])  # Exclude outliers
sorted_delays = np.sort(filtered_delays)
ccdf = 1.0 - np.arange(1, len(sorted_delays) + 1) / len(sorted_delays)

plt.figure(figsize=(10, 5))
plt.plot(sorted_delays, ccdf, marker="o", linestyle="-", color="purple")
plt.xlabel("RLC Delay (ms)")
plt.ylabel("CCDF (Log Scale)")
plt.yscale("log")  # Set Y-axis to logarithmic scale
plt.title("Logarithmic CCDF of RLC Delays (Outliers)")
plt.grid(True)
plt.show()

# 4️⃣ Plot Autocorrelation of RLC Delays Using NumPy
def autocorrelation(x, lag):
    """Compute autocorrelation for a given lag."""
    n = len(x)
    mean_x = np.mean(x)
    variance = np.var(x)
    x_shifted = np.roll(x, lag)
    autocorr = np.sum((x - mean_x) * (x_shifted - mean_x)) / (n * variance)
    return autocorr

# Compute autocorrelation for different lags
lags = np.arange(1,51 )  # Compute for 50 lags
autocorr_values = [autocorrelation(rlc_delays, lag) for lag in lags]

# Plot Autocorrelation
plt.figure(figsize=(10, 5))
plt.stem(lags, autocorr_values, basefmt=" ")
plt.xlabel("Lag")
plt.ylabel("Autocorrelation")
plt.title("Autocorrelation of RLC Delays")
plt.grid(True)
plt.show()
```

```python
import json
import numpy as np

# Load the JSON file
with open("packets.json", "r") as file:
    packets = json.load(file)

# Extract RLC delays
rlc_delays = []
for packet in packets[:40000]:  # Process first 40,000 packets
    try:
        rlc_delay = (packet['rlc.out_t'] - packet['rlc.in_t']) * 1e3  # Convert to milliseconds
        rlc_delays.append(rlc_delay)
    except KeyError:
        continue  # Skip packets with missing timestamps

# Convert to numpy array
rlc_delays = np.array(rlc_delays)

# 1️⃣ Basic Statistics
mean_delay = np.mean(rlc_delays)
median_delay = np.median(rlc_delays)
min_delay = np.min(rlc_delays)
max_delay = np.max(rlc_delays)
std_dev_delay = np.std(rlc_delays)

# 2️⃣ Percentiles
percentiles = np.percentile(rlc_delays, [99.2,99.4,99.6,99.8,99.9])

# 3️⃣ Manual Skewness Calculation
n = len(rlc_delays)
mean_diff = rlc_delays - mean_delay
skewness = np.sum(mean_diff**3) / (n * (std_dev_delay**3))

# 4️⃣ Manual Kurtosis Calculation
kurtosis = np.sum(mean_diff**4) / (n * (std_dev_delay**4))

# 5️⃣ Outlier Count (Delay > 40ms)
outlier_count = np.sum(rlc_delays > 40)

# Print Statistics
print("\n**RLC Delay Statistics** ")
print(f"Mean Delay: {mean_delay:.3f} ms")
print(f"Median Delay: {median_delay:.3f} ms")
print(f"Min Delay: {min_delay:.3f} ms")
print(f"Max Delay: {max_delay:.3f} ms")
print(f"Standard Deviation: {std_dev_delay:.3f} ms")

print("\n **Percentiles**")
print(f"99.2th Percentile: {percentiles[0]:.3f} ms")
print(f"99.4th Percentile: {percentiles[1]:.3f} ms")
print(f"99.6 Percentile: {percentiles[2]:.3f} ms")
print(f"99.8th Percentile: {percentiles[3]:.3f} ms")
print(f"99.9th Percentile: {percentiles[4]:.3f} ms")

print("\n **Distribution Analysis**")
print(f"Skewness: {skewness:.3f} (Positive = right-skewed, Negative = left-skewed)")
print(f"Kurtosis: {kurtosis:.3f} (High = many outliers)")

print(f"\n **Outliers:** {outlier_count} packets have delay > 40ms")
```

```python
#using seaborn
import json
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Read the JSON file
with open('packets.json') as file:
    data = json.load(file)

# Create lists to store entry_time, exit_time, and calculate delay
entry_list_rlc = [packet['rlc.in_t'] for packet in data]
exit_list_rlc = [packet['rlc.out_t'] for packet in data]
delay_list = [(exit_list_rlc[i] - entry_list_rlc[i]) * 1000 for i in range(len(data))]

# Filter out outliers > 40 milliseconds
filtered_delays = [delay for delay in delay_list if delay <= 40]

# Calculate the autocorrelation
autocorrelation = [np.corrcoef(filtered_delays[:-lag], filtered_delays[lag:])[0, 1] for lag in range(1, 51)]

# Create a DataFrame for Seaborn
import pandas as pd
acf_df = pd.DataFrame({'Lag': range(1, 51), 'Autocorrelation': autocorrelation})

# Plot the autocorrelation
plt.figure(figsize=(10, 5))
sns.barplot(x='Lag', y='Autocorrelation', data=acf_df, color='b')
plt.xlabel('Lag')
plt.ylabel('Autocorrelation')
plt.title('Autocorrelation of RLC Packet Delays (Filtered, <= 40ms)')
plt.grid(True)
plt.show()
```

```python
import json
import numpy as np

# Load the packets.json file
with open("packets.json", "r") as file:
    packets = json.load(file)

# Extract RLC delays and corresponding sequence numbers (SN)
rlc_delays = []
sn_values = []

for packet in packets[:40000]:  # Process first 40,000 packets
    try:
        rlc_delay = (packet['rlc.out_t'] - packet['rlc.in_t']) * 1e3  # Convert to milliseconds
        rlc_delays.append(rlc_delay)
        sn_values.append(packet["sn"])  # Store SN for reference
    except KeyError:
        continue  # Skip packets with missing timestamps

# Convert lists to numpy arrays
rlc_delays = np.array(rlc_delays)
sn_values = np.array(sn_values)

# 1️⃣ Find SN of Packet with Maximum RLC Delay
max_index = np.argmax(rlc_delays)
max_sn = sn_values[max_index]
max_delay = rlc_delays[max_index]

# 2️⃣ Find SN of Packet with Minimum RLC Delay
min_index = np.argmin(rlc_delays)
min_sn = sn_values[min_index]
min_delay = rlc_delays[min_index]

# 3️⃣ Find SN of Packet with RLC Delay Closest to 40ms
closest_40_index = np.abs(rlc_delays - 60).argmin()
closest_40_sn = sn_values[closest_40_index]
closest_40_delay = rlc_delays[closest_40_index]

# 4️⃣ Find SN of Packet with RLC Delay Closest to 80ms
closest_80_index = np.abs(rlc_delays - 80).argmin()
closest_80_sn = sn_values[closest_80_index]
closest_80_delay = rlc_delays[closest_80_index]

# Print results
print("\n📊 **RLC Delay Analysis Results** 📊")
print(f"🔹 Maximum RLC Delay: {max_delay:.3f} ms (SN: {max_sn})")
print(f"🔹 Minimum RLC Delay: {min_delay:.3f} ms (SN: {min_sn})")
print(f"🔹 Closest to 40ms: {closest_40_delay:.3f} ms (SN: {closest_40_sn})")
print(f"🔹 Closest to 80ms: {closest_80_delay:.3f} ms (SN: {closest_80_sn})")
```
